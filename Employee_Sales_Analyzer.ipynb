```python
try:
    Employees = int(input("How many employees: "))
except ValueError:
    print("Invalid employee count.")
    Employees = 0

print(Employees)

Employee_data = []

for i in range(Employees):

    name = input("Enter Your name: ")

    try:
        Sales_amount = float(input("Enter sales amount: "))
    except ValueError:
        print("Invalid sales amount.")
        continue

    Employee_data.append([name, Sales_amount])


Total_sales = sum(sales for name, sales in Employee_data)

Average_Sales = Total_sales / Employees


Highest_salesperson = max(Employee_data, key=lambda x: x[1])


if Average_Sales >= 70000:
    Performance = "Excellent"

elif Average_Sales >= 50000:
    Performance = "Good"

elif Average_Sales >= 30000:
    Performance = "Average"

else:
    Performance = "Needs Improvement"


print("\n================ Employee Sales Analyzer ================")

for name, sales in Employee_data:
    print(name, ":", sales)

print("-------------------------------------------")

print("Total Sales:", Total_sales)
print("Average Sales:", Average_Sales)
print("Highest Salesperson:", Highest_salesperson[0])
print("Highest Sales Amount:", Highest_salesperson[1])
print("Performance:", Performance)

print("===========================================")
```

```python
try:
    Employees = int(input("How many employees: "))

    Employee_data = []

    for i in range(Employees):
        name = input("Enter employee name: ")
        Sales_amount = float(input("Enter sales amount: "))

        Employee_data.append([name, Sales_amount])

    # Total sales
    Total_sales = sum(sales for name, sales in Employee_data)

    # Average sales
    Average_Sales = Total_sales / Employees

    # Highest salesperson
    Highest_salesperson = max(Employee_data, key=lambda x: x[1])

    # Performance
    if Average_Sales >= 70000:
        Performance = "Excellent"
    elif Average_Sales >= 50000:
        Performance = "Good"
    elif Average_Sales >= 30000:
        Performance = "Average"
    else:
        Performance = "Needs Improvement"

    # Final report
    print("\n========== Employee Sales Report ==========")

    for name, sales in Employee_data:
        print(name, ":", sales)

    print("-------------------------------------------")
    print("Total Sales:", Total_sales)
    print("Average Sales:", Average_Sales)
    print("Highest Salesperson:", Highest_salesperson[0])
    print("Highest Sales Amount:", Highest_salesperson[1])
    print("Performance:", Performance)

    print("===========================================")

except ValueError:
    print("Please enter a valid number.")
```
